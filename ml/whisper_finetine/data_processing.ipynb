```python
from datasets import load_dataset, Audio

# Load English dataset in streaming mode using the correct split 'en2indic'
dataset_eng = load_dataset(
    "openslr/librispeech_asr",
    "clean",
    split="train.100"
)
```

```python
# Load Nepali dataset in streaming mode
dataset_nep = load_dataset(
    "iamTangsang/OpenSLR54-Nepali-ASR",
    split="train"
)


# Filter English dataset for audio lengths <= 10 seconds (based on 'begin_time' and 'end_time' or available duration metadata)
# Note: Librispeech in datasets typically has a 'begin_time' and 'end_time' for chunks or a duration field.
# dataset_eng = dataset_eng.filter(lambda x: (x.get('end_time', 10) - x.get('begin_time', 0)) <= 10.0)



# Cast audio columns to 16kHz
dataset_eng = dataset_eng.cast_column("audio", Audio(sampling_rate=16000, decode=True))
dataset_nep = dataset_nep.cast_column("utterance", Audio(sampling_rate=16000, decode=True))

dataset_nep = dataset_nep.rename_column("utterance", "audio")
dataset_nep = dataset_nep.rename_column("transcription", "text")
```

```python
# dataset_eng = dataset_eng.remove_columns(['file', 'speaker_id', 'chapter_id', 'id'])
dataset_nep = dataset_nep.remove_columns(['utterance_id', 'speaker_id', 'num_frames'])
```

```python
print(dataset_eng)
print(dataset_nep)
```

```python
import numpy as np
import io
from datasets import Dataset, Audio, Features, Value, interleave_datasets
import soundfile as sf

repository_id = "kkarhm/nepali-english-codemixed-asr"

print(f"Preparing combined dataset for {repository_id}...")

def transform(example):
    return {
        "audio": example["audio"],
        "text": example["text"]
    }

from datasets import concatenate_datasets

dataset = concatenate_datasets([dataset_nep, dataset_eng])
final_dataset = dataset.shuffle(seed=42)



features = Features({
    "audio": Audio(sampling_rate=16000),
    "text": Value("string")
})
final_dataset = final_dataset.cast(features)

print("Uploading dataset to Hub...")
final_dataset.push_to_hub(repository_id)
print("Upload complete!")
```
